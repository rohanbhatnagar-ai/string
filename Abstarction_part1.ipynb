```python
# 16. Class Name: Shape
# Problem
# Create an abstract class Shape.
# Explanation
# Declare abstract methods
# • calculate_area()

# Sumit Sir Notes
# • calculate_perimeter()
# Create subclasses like
# • Circle
# • Rectangle
# Implement both methods

from abc import ABC, abstractmethod
import math

# abstract Class
class Shape(ABC):

    @abstractmethod
    def calculate_perimeter(self):
        pass
# subclass
class Circle(Shape):

    def __init__(self,radius):
        self.radius = radius
    def calculate_area(self):
        return math.pi*self.radius*self.radius
    def calculate_perimeter(self):
        return 2*math.pi*self.radius
# subclass
class Rectangle(Shape):
    def __init__(self, length,width):
        self.length = length
        self.width = width
    def calculate_area(self):
        return self.length*self.width
    def calculate_perimeter(self):
        return 2 * (self.length + self.width)
# object
circle = Circle(5)

print("circle Area:", circle.calculate_area())
print("circle Perimeter:",circle.calculate_perimeter())

rectangle = Rectangle(10,5)

print("Rectangle Area:",rectangle.calculate_area())
print("rectangle Perimeter:",rectangle.calculate_perimeter())
```

```python
# 17. Class Name: Vehicle
# Problem
# Create abstract class Vehicle.
# Explanation
# Abstract methods
# • start()
# • stop()
# Implement in
# • Car
# • Bike
# Each subclass should define how the vehicle starts and stops.
from abc import ABC, abstractmethod

#Abstarct class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def stop(self):
        pass

# Car class
class Car(Vehicle):
    def start(self):
        print("Car starts with a key or push button")
    def stop(self):
        print("Car stops when the driver applies the brake")
# Bike clas
class Bike(Vehicle):

    def start(self):
        print("Bike starts using self-start or kick")
    def stop(self):
        print("Bike stops when the rider applies the brakr")

# display
car = Car()
car.start()
car.stop()

print()

bike = Bike()
bike.start()
bike.stop()
```

```python
# 18. Class Name: Payment
# Problem
# Create abstract class Payment.
# Explanation
# Abstract method
# process_payment(amount)
# Create subclasses
# • CreditCardPayment
# • UPIPayment
# Each should process payment differently.
from abc import ABC, abstractmethod


#Abstract class
class Payment(ABC):

    @abstractmethod
    def process_payment(self,amount):
        pass
# Credit Card Payment
class CreditCardPayment(Payment):
    def process_payment(self,amount):
        
        print(f"{amount} payment processed  using Card")
# UPI Payment
class UPIPayment(Payment):

    def process_payment(self, amount):
        print(f"{amount} payment processed using UPI.")
# display
credit_card = CreditCardPayment()
credit_card.process_payment(500)

upi = UPIPayment()
upi.process_payment(1000)
```

```python
# 19. Class Name: Animal
# Problem
# Create abstract class Animal.
# Explanation
# Abstract method
# make_sound()
# Create subclasses
# • Dog
# • Cat
# Each returns its own sound.

from abc import ABC,abstractmethod

#abstract Class
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

# Dog Class
class Dog(Animal):
    def make_sound(self):
        return "Dog says: Woof Woof"

# Cat class
class Cat(Animal):

    def make_sound(self):
        return "Cat says: Meow Meow"

# display
dog = Dog()
print(dog.make_sound())
cat = Cat()
print(cat.make_sound())
```

```python
# 20. Class Name: Report
# Problem
# Create abstract class Report.
# Explanation
# Abstract method
# generate_report()
# Create
# • PDFReport
# • ExcelReport
# Each generates report differently.

from abc import ABC, abstractmethod

#Abstract class
class Report(ABC):

    @abstractmethod
    def generate_report(self):
        pass
# PDF Report
class PDFReport(Report):

    def generate_report(self):
        return "Report generated in pdf Format"

# Excel Report
class ExcelReport(Report):
    def generate_report(self):
        return "Report generated in Excel format"
# display
pdf = PDFReport()
print(pdf.generate_report())

excel = ExcelReport()
print(excel.generate_report())
```

```python
# 21. Class Name: BankAccount
# Problem
# Create abstract class BankAccount.
# Explanation
# Abstract method

# Sumit Sir Notes
# calculate_interest()
# Implement in
# • SavingsAccount
# • CurrentAccount
# Each account type should use a different interest calculation.



from abc import ABC , abstractmethod

#abstract Class
class BankAccount(ABC):

    @abstractmethod
    def calculate_interest(self):
        pass
# Savings Account
class SavingsAccount(BankAccount):
    def __init__(self, balance):
        self.balance = balance

    def calculate_interest(self):
        interest = self.balance * 0.04
        return interest

# Current Account
class CurrentAccount(BankAccount):

    def __init__(self,balance):
        self.balance = balance
    def calculate_interest(self):
        interest = self.balance * 0.01
        return interest
# display
savings =  SavingsAccount(100000)
print("Savings Account Interest:", savings.calculate_interest())

current = CurrentAccount(100000)
print("Current Account Interest:",current.calculate_interest())
```

```python
# 23. Class Name: Calculator
# Problem
# Create abstract class Calculator.
# Explanation
# Abstract method
# compute(a, b)
# Implement in
# • AddCalculator
# • MultiplyCalculator



from abc import ABC, abstractmethod

#abstract Class


class Calculator(ABC):

    @abstractmethod
    def compute(self, a, b):
        pass

# Addition Calculator
class AddCalculator(Calculator):

    def compute(self, a, b):
        return a+b
# Multiplication Calculator
class MultiplyCalculator(Calculator):

    def compute(self, a,b):
        return a * b

#display
add = AddCalculator()
print("Addition:",add.compute(10,5))

multiply = MultiplyCalculator()
print("Multiplication:", multiply.compute(10,5))

```

```python
# 27. Class Name: Database
# Problem
# Create abstract class Database.
# Explanation
# Abstract methods
# • connect()
# • query(sql)
# Implement in
# • MySQLDatabase
# • MongoDatabase
# Each should define how it connects and executes queries.

from abc import ABC, abstractmethod

# Abstract class
class Database(ABC):
    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def query(self, sql):
        pass
# MYSQL Databsse
class MySQLDatabase(Database):

    def connect(self):
        print("Connected to MySQL database")
    def query(self,sql):
        print(f"sql excecuting query: {sql}")
# MongoDB Database
class MongoDatabase(Database):
    def connect(self):
        print("connected to MongoDB database")
    def query(self,sql):
        print(f"MongoDB exceuting query:{sql}")
#MySQL
mysql = MySQLDatabase()
mysql.connect()
mysql.query("SELECT * FROM employees")

print()

# MongoDB 
mongo = MongoDatabase()
mongo.connect()
mongo.query("Find all employees")
        
```
