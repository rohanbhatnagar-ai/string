```python
# 1. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# A shape can have different formulas for area and perimeter. Therefore, create two abstract
# methods:
# • area()
# • perimeter()
# Create subclasses like Circle and Rectangle and implement both methods

from abc import ABC, abstractmethod
import math

# abstract Class
class Shape(ABC):

    @abstractmethod

    def area(self):
        pass
        
# Circle Class
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius

    def perimeter(self):
        return 2 * math.pi * self.radius

# Rectangle Class
class Rectangle(Shape):

    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length*self.width
    def perimeter(self):
        return 2* (self.length + self.width)
# Circle object
circle = Circle(5)
print("Circle Perimeter:", circle.area())
print("Circle Perimeter:", circle.perimeter())

print()

# display
rectangle = Rectangle(10,5)
print("area:",rectangle.area())
print("Reactangle perimeter:",rectangle.perimeter())

        
```

```python
# 2. Class Name: Vehicle
# Problem
# Create an abstract class named Vehicle.
# Explanation
# Every vehicle can start and stop, but each vehicle does it differently.
# Create two abstract methods:
# • start()
# • stop()
# Create a Car class that implements both methods.
from abc import ABC , abstractmethod

# abstract class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def stop(self):
        pass
# car Class
class Car(Vehicle):

     def start(self):
         print("car starts with a key or push bottom")
     def stop(self):
         print("Car stops when the driver applies the brake")
# display
car = Car()

car.start()
car.stop()

```

```python
# 3. Class Name: Animal
# Problem
# Create an abstract class named Animal.
# Explanation
# Every animal makes a sound, but the sound depends on the animal.

# Sumit Sir Notes
# Create an abstract method:
# • make_sound()
# Create subclasses:
# • Dog
# • Cat
# Each should implement its own sound.
from abc import ABC, abstractmethod
#abstract Class

class Animal(ABC):
    def make_sound(self):
        pass
# Dog class
class Dog(Animal):

    def make_sound(self):
        return "Dog says:WOOF WOOF"
    # cat Class
class Cat(Animal):
        
    def make_sound(self):
        return "Cat says:Meow Meoww"
#display
dog = Dog()
print(dog.make_sound())
cat = Cat()

print(cat.make_sound())


```

```python
# 4. Class Name: Payment
# Problem
# Create an abstract class named Payment.
# Explanation
# Every payment processes money differently.
# Create an abstract method:
# • process_payment(amount)
# Create subclasses:
# • CreditCard
# • Cash
# Each should process the payment in its own way.



from abc import ABC , abstractmethod

#abstractmethod
class Payment(ABC):

    @abstractmethod
    def process_payment(self,amount):
        pass
# CreditCard Class
class CreditCard(Payment):

    def process_payment(self,amount):
        return f"${amount} payment processed using cash"

# Cash Class
class Cash(Payment):

    def process_payment(self, amount):
        return f"₹{amount} payment processed using Cash."


# create objects
credit_card = CreditCard()
print(credit_card.process_payment(500))

cash = Cash()

print(cash.process_payment(1000))
```
