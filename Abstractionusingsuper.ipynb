```python
# 16. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# The constructor should accept a shape's color.
# __init__(self, color)
# Create an abstract method:
# • area()
# Create subclass:
# • Circle
# Use super() to initialize the color.



from abc import ABC, abstractmethod
import math

#abstract Class
class Shape(ABC):

    def __init__(self,color):
        self.color = color

    @abstractmethod
    def area(self):
        pass
# Circle Class
class Circle(Shape):

    def __init__(self,color,radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return math.pi*self.radius * self.radius
# display
circle = Circle("Red", 5)

print("Color:",circle.color)
print("Area:", circle.area())
```

```python
# 17. Class Name: Vehicle
# Problem
# Create an abstract class named Vehicle.
# Explanation
# Store the vehicle brand in the constructor.
# __init__(self, brand)
# Create abstract methods:
# • start()
# • stop()
# Create subclass:

# Sumit Sir Notes
# • Car
# Use super() to initialize the brand.

from abc import ABC, abstractmethod

# abstract Class
class Vehicle(ABC):

    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# Car Class
class Car(Vehicle):
    def __init__(self,brand):
        super().__init__(brand)

    def start(self):
        return "Car starts eith a key or push button"

    def stop(self):
        return "Car stops when the braake is applied"
#display
car = Car("TOYOTA")


print("Brand:", car.brand)
print(car.start())
print(car.stop())
```

```python
# 21. Class Name: Report
# Problem
# Create an abstract class named Report.
# Explanation
# Store the report title.
# __init__(self, title)
# Create abstract method:
# • generate()
# Create subclass:
# • PDFReport
# Use super() to initialize the title.

from abc import ABC, abstractmethod

#abstract Class
class Report(ABC):

    def __init__(self,title):
        self.title = title
    @abstractmethod
    def generate(self):
        pass
#PDFReport Class
class PDFReport(Report):
    def __init__(self,title):
        super().__init__(title)
    def generate(self):
        return f"PDF Report Generated: {self.title}"
#display
report = PDFReport("Monthly Sales Report")

print("Report Title:",report.title)
print(report.generate())
```

```python
# 24. Class Name: Transport
# Problem
# Create an abstract class named Transport.
# Explanation
# Store the transport speed.
# __init__(self, speed)
# Create abstract methods:

# Sumit Sir Notes
# • move()
# • fuel_type()
# Create subclass:
# • Car
# Use super() to initialize the speed.


from abc import ABC, abstractmethod

#abstract Class
class Transport(ABC):


    def __init__(self,speed):
        self.speed = speed
    @abstractmethod
    def fuel_type(self):
        pass
# Car Class
class Car(Transport):

    def __init__(self,speed):
        super().__init__(speed)
    def move(self):
        return f"Car is moving at{self.speed} km/h"
    def fuel_type(self):
        return "Car uses Petrol"

# display
car = Car(80)

print("Speed:", car.speed)

print(car.move())
print(car.fuel_type())
```

```python
# 25. Class Name: Notification
# Problem
# Create an abstract class named Notification.
# Explanation
# Store the recipient name.
# __init__(self, recipient)
# Create abstract method:
# • send(message)
# Create subclass:
# • EmailNotification
# Use super() to initialize the recipient.

from abc import ABC, abstractmethod
#abract Class
class Notification(ABC):

    def __init__(self, recipient):
        self.recipient = recipient

    @abstractmethod
    def send(self, message):
        pass

# EmailNotifications Class
class EmailNotification(Notification):
    def __init__(self, recipient):
        super().__init__(recipient)
    def send(self, message):
        return f"Email sent to {self.recipient}: {message}"
# display
email= EmailNotification("Rohan")

print("Recipient:",email.recipient)
print(email.send("Your interview is scheduled tommorrow"))
```

```python
# 27. Class Name: Employee
# Problem
# Create an abstract class named Employee.
# Explanation
# Store:
# • Employee name
# • Base salary
# Create abstract method:
# • calculate_bonus()
# Create subclass:
# • Developer
# Use super() to initialize the common employee information before calculating the bonus.

from abc import ABC, abstractmethod

#abstract Class
class Employee(ABC):

    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary
    @abstractmethod
    def calculate_bonus(self):
        pass
# developer Class
class Developer(Employee):

    def __init__(self, name, base_salary):
        super().__init__(name, base_salary)
    def calculate_bonus(self):
        bonus = self.base_salary * 0.10
        return bonus

#display
developer = Developer("Rohan",50000)

print("Employee Name:", developer.name)
print("Base Salary:", developer.base_salary)
print("Bonus:", developer.calculate_bonus())
```
