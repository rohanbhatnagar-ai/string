```python
# 1. Person → Student, Teacher
# Question
# Create a parent class Person with attributes name and age. Create two child classes Student
# and Teacher that inherit from Person.
# Concept
# • Hierarchical Inheritance
# • Parent attributes inherited by multiple child classes.
# How to Solve
# 1. Create a parent class Person.
# 2. Add name and age using a constructor.
# 3. Create Student(Person).
# 4. Create Teacher(Person).
# 5. Create objects of both classes.
# 6. Show that both children can access name and age.


class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

class Student(Person):
    def study(self):
        print(self.name, "is studying")
class Teacher(Person):
    def teach(self):
        print(self.name, "is teaching")

# display
student1 = Student("Rohan",21)
#create teacher
teacher1 = Teacher("Amit", 35)

print("Student name:",student1.name)
print("Student Age:",student1.age)
student1.study()

```

```python
# 2. Animal → Dog, Cat
# Question
# Define class Animal with method eat(). Create subclasses Dog and Cat.
# Concept
# • Method inheritance.
# How to Solve
# 1. Create parent class Animal.
# 2. Write method eat().
# 3. Create child classes Dog and Cat.
# 4. Create objects.
# 5. Call eat() using both objects.


class Animal:
    def eat(self):
        print("Animal is eating")
class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Cat(Animal):
    def meow(self):
        print("Cat is meowing")
# display
dog1 = Dog()
cat1 = Cat()

# call inherited eat() method
dog1.eat()
cat1.eat()
# call their own mwthods
dog1.bark()
cat1.meow()
```

```python
# 3. Vehicle → Car, Bike
# Question
# Create class Vehicle with method start(). Create subclasses Car and Bike.
# Concept
# • One parent method shared by multiple children.
# How to Solve
# 1. Create Vehicle.
# 2. Add start().
# 3. Inherit it in Car.
# 4. Inherit it in Bike.
# 5. Call start() from both child objects.




class Vehicle:
    def start(self):
        
        print("Vehicle is starting")

class Car(Vehicle):
    def drive(self):
    
        print("Car is driving")
class Bike(Vehicle):
    def ride(self):
        print("Car is driving")

# display
car1 = Car()
bike1 = Bike()

#call inherited start() methodes
car1.start()
bike1.start()

# call their own methodes
car1.drive()
bike1.ride()
```

```python
# 5. Employee → Manager, Developer
# Question

# Sumit Sir Notes
# Create class Employee with attributes emp_id and salary. Create subclasses Manager and
# Developer.
# Concept
# • Attribute inheritance.
# How to Solve
# 1. Create parent constructor.
# 2. Store emp_id and salary.
# 3. Create two child classes.
# 4. Show inherited attributes.


class Employee:
    def __init__(self, emp_id,salary):
        self.emp_id = emp_id
        self.salary = salary
class Manager(Employee):
    def manage_team(self):
        print("Manager is managing the team")

class Developer(Employee):
    def write_code(self):
        print("Developer is writing code")

# create the manager 
manager1 = Manager(101, 7000)

# create devloper object
developer1= Developer(102,60000)

# display inherited attributes of Manager
print("Manager")
print("Employee ID:",manager1.emp_id)
print("Salary:",manager1.salary)
manager1.manage_team()

print()

# display inherited attributes of developer
print("Developer")
print("Employee ID:", developer1.emp_id)
developer1.write_code()
```

```python
#Hierarchical Inheritance with Constructors & Methods
# 13. Person Constructor
# Concept
# Using super() to call the parent constructor.
# How to Solve
# 1. Create constructor in parent.
# 2. Create child constructors.
# 3. Use super().__init__().
# 4. Add child-specific attribute.
# 5. Display all data.


class Person:
    def __init__(self, name,age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:",self.marks)
# display
student1 = Student("Rohan", 21, 85)

student1.display()
```

```python
# 14. Animal Sound Override
# Concept
# Method Overriding.
# How to Solve
# 1. Parent has sound().
# 2. Dog overrides.
# 3. Cat overrides.
# 4. Call method from each object.

class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meows")

#display
animal1 = Animal()
dog1 = Dog()
cat1 = Cat()

# call sound() methodes
animal1.sound()
dog1.sound()
cat1.sound()
```

```python
# 18. BankAccount Interest
# Concept
# Different child behavior.
# How to Solve
# 1. Parent constructor.
# 2. Savings has one interest.
# 3. Current has another.

# Sumit Sir Notes
# 4. Calculate separately.

class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def calculate_interest(self):
        print("Interest calculation")
class SavingsAccount(BankAccount):
    def calculate_interest(self):
        interest = self.balance * 5 / 100
        print("Savings Account Interest:", interest)

class CurrentAccount(BankAccount):
    def calculate_interest(self):
        interest = self.balance * 2/ 100
        print("Current Account Intrest:", interest)

# display
savings = SavingsAccount(10000)
current = CurrentAccount(10000)

# savings
savings.calculate_interest()
current.calculate_interest()
```

```python
# 19. Device Info
# Concept
# Method overriding.
# How to Solve
# 1. Parent method info().
# 2. Phone overrides.
# 3. Tablet overrides.


class Device:
    def info(self):
        print("This is a device")

class Phone(Device):
    def info(self):
        print("This is a smartphone")

class Tablet(Device):
    def info(self):
        print("This is a tablet")

# display
device1 = Device()
phone1 = Phone()
tablet1 = Tablet()

# call info() methodes
device1.info()
phone1.info()
tablet1.info()
```

```python
# 20. Game Start
# Concept
# Inheritance + Additional Methods.
# How to Solve
# 1. Parent method start().
# 2. Chess inherits.
# 3. Chess adds checkmate().
# 4. Ludo adds roll_dice().


class Game:
    def start(self):
        print("Chess: Checkmate:")
class Chess(Game):
    def checkmate(self):
        print("Chess: Checkmate")
class Ludo(Game):
    def roll_dice(self):
        print("Ludo; Dice rolled")

#display
chess1 = Chess()
ludo1 = Ludo()

chess1.start()
ludo1.start()
```
