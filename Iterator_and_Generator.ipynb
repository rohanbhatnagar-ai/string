```python
# dic
dicts = {"name":"sumit",
         "Age":12,
         "Gender":"Male",
         "State":"TN"
        }
iterator_key = iter(dicts.key())
iterator_values = iter(dicts.values())
print(f"next{(iterator_key)}: value ={next(iterator_values)}")
print(f"next{(iterator_key)}: value ={next(iterator_values)}")
print(f"next{(iterator_key)}: value ={next(iterator_values)}")

```

```python
# # generator 
def generator_name():

    yield value1
    yield value2
    yield value3

g = generator_name()

print(next(g))
print(next(g))
print(next(g))

```

```python
def lists(num):
    for i in num:
        return i
lists([1,2,3])
```

```python
def lists():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5
    yield 6

generator = lists()
print(next(generator))
print(next(generator))
print(next(generator))
print(next(generator))
print(next(generator))

```

```python
def number():
    for i in range(1,7):
        yield (i)
generator = number()
for i in generator:
    print(i)
```
