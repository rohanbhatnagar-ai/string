```python
# 1. Class Name: Person → Student
# Problem
# Create a parent class named Person and a child class named Student.
# Explanation
# The Person class should store:
# • name
# • age
# The Student class should inherit from Person and add:
# • marks
# Create a method in Student to display the student's name, age, and marks.





class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.marks = marks
    def display(self):
        print("Name:",self.name)
        print("Age:", self.age)
        print("Marks:",self.marks)
#display
student1=Student("Rohan", 21,85)

student1.display()
```

```python
# 2. Class Name: Vehicle → Car
# Problem
# Create a parent class named Vehicle.
# Explanation
# The Vehicle class should contain a method:
# • start()
# Create a child class Car that:
# • Adds a brand attribute.
# • Overrides the start() method to display a car-specific starting message.


class Vehicle:
    def start(self):
        print("Vehicle is starting")

class Car(Vehicle):
    def __init__(self,brand):
        self.brand =  brand
    # Method overriding
    def start(self):
        print(f"{self.brand} car is starting")
# create Car object
car1 = Car("audi")
car1.start()
```

```python
# 3. Class Name: Animal → Dog
# Problem
# Create a parent class named Animal.
# Explanation

# Sumit Sir Notes
# The Animal class should contain:
# • eat()
# Create a child class Dog that:
# • Inherits from Animal.
# • Adds a method:
# o bark()
# Demonstrate both methods.


class Animal:
    def eat(self):
        print("Animal is eating")
class Dog(Animal):
    def bark(self):
        print("dog is barking")

# display
dog1=Dog()
#demonstrate both methods
dog1.eat()
dog1.bark()
```

```python
# 4. Class Name: Employee → Manager
# Problem
# Create a parent class named Employee.
# Explanation
# The Employee class should store:
# • emp_id
# • salary
# Create child class Manager.
# Add:
# • department
# Display all employee information.




class Employee:
    def __init__(self,emp_id,salary):
        self.emp_id = emp_id
        self.salary = salary

class Manager(Employee):
    def __init__(self,emp_id,salary,department):
        super().__init__(emp_id,salary)
        self.department = department
    def display(self):
        print("emp_id:",self.emp_id)
        print("salary:",self.salary)
        print("department:",self.department)
#display()
empploy =Manager(1234,40000,"CSE")
empploy.display()
```

```python
# 5. Class Name: Shape → Circle
# Problem
# Create a parent class named Shape.
# Explanation
# Create a method:
# • area()
# Create child class:
# • Circle

# Sumit Sir Notes
# Store:
# • radius
# Override area() to calculate the area of a circle.

class Shape:
    def area(self):
        print("Area od shape")
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    # Method overrding
    def area(self):
        result = 3.14 * self.radius * self.radius
        print("Area of Circle:",result)
#display
circle1 = Circle(5)

circle1.area()
```
