```python
# Encapsulation Practice Questions (1–15)
# 1. Class Name: BankAccount
# Problem
# Create a class named BankAccount.
# Explanation
# A bank account stores money. The balance should not be directly accessible from outside
# the class, so create a private attribute named __balance.
# Implement the following methods:
# • deposit(amount)
# o Deposit money into the account.
# o Amount must be greater than 0.
# • withdraw(amount)
# o Withdraw money.
# o Withdrawal should happen only if:
# ▪ amount > 0
# ▪ amount <= available balance



class BankAccounts:

    def __init__(self,balance):
        self._balance = balance
    # setter
    def deposit(self,amount):
        if amount>0:
            self._balance += amount
            print("Amount:",amount)
        else:
            print("amount should be greater than zero")
    def Withdraw(self,amount):
        if amount>0 and amount< self._balance:
            self._balance -= amount
            print("withdrawn amount",amount)
        else:
            print("does not amount")

    def get_balance(self):
        return get_balance
# details
account = BankAccounts(80000)
# deposits 
account.deposit(2000)
# withdraw
account.Withdraw(1000)



            
        
    
        
    
```

```python
# 14. Class Name: Laptop
# Problem
# Create class Laptop.
# Explanation
# Private attributes
# • __ram
# • __storage
# Provide
# • getter
# • setter

# Sumit Sir Notes
# Validation
# RAM and Storage should always be positive.
class Laptop:

    def __init__(self, ram, storage):
        self.__ram = ram
        self.__storage = storage

    # RAM getter
    def get_ram(self):
        return self.__ram

    # RAM setter
    def set_ram(self, ram):
        if ram > 0:
            self.__ram = ram
            print("RAM updated successfully")
        else:
            print("RAM should be positive")

    # Storage getter
    def get_storage(self):
        return self.__storage

    # Storage setter
    def set_storage(self, storage):
        if storage > 0:
            self.__storage = storage
            print("Storage updated successfully")
        else:
            print("Storage should be positive")


# Creating object
l1 = Laptop(8, 512)

# Get values
print("RAM:", l1.get_ram())
print("Storage:", l1.get_storage())

# Update values
l1.set_ram(16)
l1.set_storage(1024)

# Get updated values
print("RAM:", l1.get_ram())
print("Storage:", l1.get_storage())

# Validation
l1.set_ram(-8)
l1.set_storage(0)
```
