```python
from abc import ABC, abstractmethod

# ===========================
# Custom Exceptions
# ===========================

class InsufficientBalanceError(Exception):
    pass


class InvalidAmountError(Exception):
    pass


class AccountNotFoundError(Exception):
    pass


class DailyLimitExceededError(Exception):
    pass


# ===========================
# Abstract Class
# ===========================

class Account(ABC):

    total_accounts = 0
    account_number = 1001

    def __init__(self, name, balance):
        self._account_no = Account.account_number
        Account.account_number += 1
        Account.total_accounts += 1

        self._name = name
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise InvalidAmountError("Balance cannot be negative.")
        self._balance = amount

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def get_balance(self):
        pass

    def show_account_info(self):
        print("\n------ Account Info ------")
        print("Account Number :", self._account_no)
        print("Name           :", self._name)
        print("Balance        : ₹", self._balance)

    @classmethod
    def show_total_accounts(cls):
        print("\nTotal Accounts Created :", cls.total_accounts)


# saving 

class SavingsAccount(Account):

    INTEREST = 0.04
    MIN_BALANCE = 1000
    DAILY_LIMIT = 25000

    def __init__(self, name, balance):
        super().__init__(name, balance)
        self.__daily_withdrawn = 0

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0.")

        self.balance = self.balance + amount
        print("Deposit Successful.")

    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Invalid withdrawal amount.")

        if self.__daily_withdrawn + amount > SavingsAccount.DAILY_LIMIT:
            raise DailyLimitExceededError("Daily withdrawal limit exceeded.")

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance.")

        if self.balance - amount < SavingsAccount.MIN_BALANCE:
            raise InsufficientBalanceError(
                "Minimum balance of ₹1000 must be maintained."
            )

        self.balance = self.balance - amount
        self.__daily_withdrawn += amount
        print("Withdrawal Successful.")

    def get_balance(self):
        return self.balance

# creat

class CurrentAccount(Account):

    OVERDRAFT_LIMIT = 10000

    def __init__(self, name, balance):
        super().__init__(name, balance)

    def deposit(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0.")

        self.balance = self.balance + amount
        print("Deposit Successful.")

    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Invalid withdrawal amount.")

        if self.balance - amount < -CurrentAccount.OVERDRAFT_LIMIT:
            raise InsufficientBalanceError("Overdraft limit exceeded.")

        self.balance = self.balance - amount
        print("Withdrawal Successful.")

    def get_balance(self):
        return self.balance

# class 

class Bank:

    def __init__(self):
        self.accounts = {}

    def create_account(self, acc_type, name, balance):

        if acc_type == 1:
            account = SavingsAccount(name, balance)

        else:
            account = CurrentAccount(name, balance)

        self.accounts[account._account_no] = account

        print("\nAccount Created Successfully.")
        print("Account Number :", account._account_no)

    def find_account(self, account_no):

        if account_no not in self.accounts:
            raise AccountNotFoundError("Account not found.")

        return self.accounts[account_no]

############################
# Main Program
##############################

bank = Bank()

while True:

    print("\n========= BANK MENU =========")
    print("1. Create Savings Account")
    print("2. Create Current Account")
    print("3. Deposit")
    print("4. Withdraw")
    print("5. Check Balance")
    print("6. Show Account Info")
    print("7. Show Total Accounts")
    print("8. Exit")

    choice = int(input("Enter Choice : "))

    try:

        if choice == 1:

            name = input("Enter Name : ")
            balance = float(input("Enter Initial Balance : "))
            bank.create_account(1, name, balance)

        elif choice == 2:

            name = input("Enter Name : ")
            balance = float(input("Enter Initial Balance : "))
            bank.create_account(2, name, balance)

        elif choice == 3:

            acc = int(input("Account Number : "))
            amount = float(input("Deposit Amount : "))

            account = bank.find_account(acc)
            account.deposit(amount)

        elif choice == 4:

            acc = int(input("Account Number : "))
            amount = float(input("Withdraw Amount : "))

            account = bank.find_account(acc)
            account.withdraw(amount)

        elif choice == 5:

            acc = int(input("Account Number : "))
            account = bank.find_account(acc)

            print("Balance : ₹", account.get_balance())

        elif choice == 6:

            acc = int(input("Account Number : "))
            account = bank.find_account(acc)

            account.show_account_info()

        elif choice == 7:

            Account.show_total_accounts()

        elif choice == 8:

            print("Thank You...")
            break

        else:
            print("Invalid Choice.")

    except (InvalidAmountError,
            InsufficientBalanceError,
            DailyLimitExceededError,
            AccountNotFoundError) as e:

        print("Error :", e)

    except Exception as e:

        print("Unexpected Error :", e)

    else:

        print("Operation Completed Successfully.")

    finally:

        print("------------------------------")
```
