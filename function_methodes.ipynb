```python
# 13-18: Positional Arguments & Keyword Arguments
# 13. Write a function full_name(first, last) and call it using positional arguments.
def full_name(first,last):
    print(first)
    print(last)
full_name("sumit","sartale")
```

```python
# 14. Call the same function using keyword arguments.
def full_name(first, last):
    print(first)
    print(last)
full_name(first="sumit",last="sir")
```

```python
# 15. Write a function details(name, age, city) and demonstrate both positional and keyword
# calls.
# positional
def function_details(name,age,city):
    print(name)
    print(age)
    print(city)
function_details("sumit","25","hyderabad")

# keyword 
def function_details(name,age,city):
    print(name)
    print(age)
    print(city)
function_details(name ="sumit sir",age="25",city="hyderabad")
```

```python
# *19-25: args (Multiple Positional Arguments)
# 19. Write a function add_all(*args) that returns sum of all numbers passed.
def add(*args):
    return sum(args)
add(10,20,30)
```

```python
# 20. Create a function print_all(*args) that prints all arguments

def prints_all(*args):
    for value in args:
        print(value)
print(12,"python",2.5)
```

```python
# 21. Write a function find_max(*args) that returns the maximum value
def find_max(*args):
    return max(args)
find_max(10,20)
```

```python
#22. Make a function concatenate(*args) that joins all string arguments with space
def concatenate(*args):
    return " ".join(args)
print(concatenate("python","is","tough language"))
```
