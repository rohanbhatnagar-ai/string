```python
# List Comprehension with Conditions (11-20)
# 11.Write a list comprehension to get numbers divisible by 3 between 1 and 30.

result = [x for x  in range(1,31) if x%3==0]

print(result)

```

```python
# 12.Filter all odd numbers from 1 to 15 using list comprehension.
odd = [x for x in range(1,16) if x%2 !=0]
print(odd)
```

```python
# 13.Use if-else in list comprehension to label numbers as "Even" or "Odd" for 1 to
# 8.
result =["Even" if x%2 == 0 else "odd" for x in range(1,9)]
print(result)
```

```python
#14.Create a list of squares of only even numbers from 1 to 10.
result = [x**2 for x in range(1,11) if x%2 ==0]
print(result)
```

```python
# 15.What will be the output: [x for x in range(10) if x % 2 == 0]
result = [ x for x in range(10) if x%2 == 0]
print(result)
```

```python
# 16.Write a list comprehension to remove vowels from a string "Python
# Programming".
text = "python programming"
results = [char for char in text if char.lower() not in "aeiou"]
print(" ".join(results))
```

```python
# 17.Get all words with length greater than 4 from a list of words.
words = ["python" , "java","data","programming"," ai ","machine"]
result = [word for word in  words if len(word)  > 4]
print(result)
```

```python
#18.Use list comprehension to create a list of tuples: [(i, i**2) for i in range(5)]
result = [(i, i**2) for i in range(5)]
print(result)
```

```python
# 21.Create a multiplication table (1 to 5) using nested list comprehension.
tables = [[i*j for j in range(1,11)] for i in range(1,6)]
print(tables)
```

```python
# 22.Flatten this nested list using list comprehension: [[1,2],[3,4],[5,6]]
numbers = [[1,2],[3,4],[5,6]]

result=[x for row in numbers for x in row]
result
```

```python
# 23.Write a list comprehension to get all characters from a list of strings.
words = ["python", "ai","data"]
characters = [char for word in words for char in word]
print(characters)
```

```python
# 24.Create a 3x3 matrix (numbers 1 to 9) using nested list comprehension.
matrix = [[3*i+j+1 for j in range(3)] for i in range(3)]
print(matrix)
```
