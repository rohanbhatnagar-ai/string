#Basic Multiple Inheritance

```python
# 1. Create two parent classes Father and Mother. Create a child class Child that inherits from
# both.

class Father:
    def father_method(self):
        print("This is Father's method")
class Mother:
    def mother_method(self):
        print("This is Mother's method")
class Child(Father, Mother):
    def child_method(self):
        print("This is Child method")
#display
child1 = Child()

child1.father_method()
child1.mother_method()
child1.child_method()
```

```python
# 2. Define class A with method show() and class B with method display(). Create class C that
# inherits from both and calls both methods.


class A:
    def show(self):
        print("This is method from classs A")
class B:
    def display(self):
        print("This is method from class B")
class C(A, B):
    pass

#display

m1 = C()
m1.show()
m1.display()
```

```python
# 3. Create classes Teacher and Author. Create class Person that inherits from both and has
# attributes from both.

class Teacher:
    def __init__(self, subject):
        self.subject = subject

class Author:
    def __init__(self, book):
        self.book = book

class Person(Teacher, Author):
    def __init__(self, subject, book):
        Teacher.__init__(self, subject)
        Author.__init__(self, book)
    def display(self):
        print("Subject:", self.subject)
        print("Book:",self.book)

# display
person1 = Person("Python", "Python Programming")

person1.display()
```

```python
# 4. Define class Sports and class Academic. Create class Student that inherits from both.

class Sports:
    def sports_activity(self):
        print("Student plays cricket")
class Academic:
    def study(self):
        print("Student studies Python")
class Student(Sports, Academic):
    pass

# display
student1 = Student()

student1.sports_activity()
student1.study()
```

#Multiple Inheritance with Static Methods

```python
# 11. Create two parent classes with static methods. Create a child class that inherits from
# both and calls the static methods.

class Father:
    @staticmethod
    def father_method():
        print("Father's static method")

class Mother:
    @staticmethod
    def mother_method():
        print("Mother's static method")
class Child(Father, Mother):
    pass
# display
child1 = Child()
child1.father_method()

child1.mother_method()
```

```python
# 12. Define class MathUtils with static method add(). Create class AdvancedMath that
# inherits and uses the static method.

class MathUtils:
    @staticmethod
    def add(a,b):
        return a+b
class AdvanceMath(MathUtils):
    pass
#display
math = AdvanceMath()

result = math.add(10,20)

print("Addition:", result)
```

```python
# 13. Create class Validator with static method is_valid(). Create class Form that inherits and
# uses it.

class Validator:
    @staticmethod
    def is_valid(value):
        if value != "":
            return True
        return False
class Form(Validator):
    pass
#display
form1 = Form()
result = form1.is_valid("Rohan")

print("Is valid:", result)
```

```python
# 17. Create class Calculator with static method and another class Display. Create child class
# that uses both.

class Calculator:
    @staticmethod
    def add(a, b):
        return a+b
class Display:
    def show(self, result):
        print("Result:",result)
class Child(Calculator, Display):
    pass
# display
m1 = Child()
result = m1.add(10,20)

m1.show(result)
```

#Multiple Inheritance with Abstract Methods

```python
# 19. Create two abstract classes Shape and Color. Create a concrete class that inherits from
# both and implements abstract methods.

from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Color(ABC):
    @abstractmethod
    def get_color(self):
        pass


class Circle(Shape, Color):
    def __init__(self, radius, color):
        self.radius = radius
        self.color = color

    def area(self):
        return 3.14 * self.radius * self.radius

    def get_color(self):
        return self.color


# Create object of concrete class
circle = Circle(5, "Red")

print("Area:", circle.area())
print("Color:", circle.get_color())
```

```python
# 20. Define abstract class Payment and abstract class Notification. Create class
# OnlinePayment that inherits from both.

from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class Notification(ABC):
    @abstractmethod
    def send_notification(self):
        pass


class OnlinePayment(Payment, Notification):

    def pay(self, amount):
        print("Payment of ₹", amount, "successful")

    def send_notification(self):
        print("Payment notification sent")


# Create object of concrete class
payment = OnlinePayment()

# Call implemented methods
payment.pay(500)
payment.send_notification()
```
