```python

# 1. Student Class
# Create a Student class with
# • name
# • age
# • marks
# Create two student objects and print all details using a method.
class Student:
    def __init__(self,name,age,markes):
        self.name = name
        self.age = age
        self.markes = markes

    def display(self):
        print("name",self.name)
        print("age",self.age)
        print("markes",self.markes)
        print()

    # object
student1=Student("rohan",20,88)
student2=Student("Aman",21,92)

    # display details
student1.display()
student2.display()
    
```

```python
# 2. Employee Salary
# Create an Employee class.
# Attributes
# • emp_id
# • name
# • salary
# Create a method that increases salary by ₹5000 and prints updated salary.
class Employee:
    def __init__(self,emp_ids,name,salary):
        self.emp_ids = emp_ids
        self.name = name
        self.salary = salary
    def display(self):
        print("employee:",self.emp_ids)
        print("name:",self.name)
        print("salary:",self.salary)
        print()

# object
employee1 = Employee(101,"rohan",90000)
employee2 = Employee(102,"Aman",100000)
# display details
employee1.display()
employee2.display()
```

```python
# 6. Car Speed
# Attributes
# • brand
# • speed
# Method
# increase_speed(value)
# Maximum speed should never exceed 200 km/h.

class Car:
    def __init__(self,brand,speed):
        self.brand = brand
        self.speed = speed
      # method
    def increase_speed(self,value):
        self.speed +=value
      # maximum speed should 200
        if self.speed>200:
             
            self.speed = 200
    def display(self):
        print("brand:",self.brand)
        print("speed:",self.speed)
# object display
car1= Car("BMW",180)
car1.display()
print("after increasing speed:")
car1.increase_speed(30)
print("after increasing the speed")
car1.display()
```

```python
# 23. Create class Student with private __marks. Add getter and setter methods.
class Student:
    def __init__(self,marks):
        self.__marks=marks
    # get the methods
    def get_marks(self):
        return self.marks
    # sett the methods
    def set_marks(self,marks):
        self.marks = marks
# object 
s1=Student(72)
# get the marks
print("marks:",s1.get_marks)
# sett the marks
s1.set_marks(85)
print("marks:",s1._Student__marks)
```

```python
#24. In class BankAccount, add getter and setter for private __balance with validation.
class Bankaccount:
    def __init__(self,balance):
        self.__balance = balance
    # get the method
    def get_balance(self):
        return get_balance
    # sett the method
    def set_balance(self,balance):
        self.__balance = balance
# create
bank1 = Bankaccount(3244)
# get the class
print("balance:",bank1.get_balance)
#set the valid
bank1.set_balance(3234)
# access the private
print("bankaccount",bank1._Bankaccount__balance)
```

```python
# 25. Create class Employee with private __salary. Setter should not allow value below
# 5000.

class Employee:
    def __init__(self,salarys):
        self.__salarys = salarys
    # get the method
    def get_salary(self):
        return get_salary
    # sett
    def set_salary(self,salarys):
        if salarys>=5000:
            self.salarys=salarys
            print("salary updated sucessfully")
        else:
            print("invalid salarys")
# creatings
employee1 = Employee(5000)
# access the private value
print("salarys of manager:",employee1._Employee__salarys)
# set the value
employee1.set_salary(6000)
```
