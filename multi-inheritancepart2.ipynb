```python
# 16. Class Name:
# Person → Employee → Manager
# Problem
# Create a three-level inheritance hierarchy.
# Explanation
# Person
# Store:
# • name
# ↓
# Employee
# Add:
# • employee_id
# ↓
# Manager
# Add:
# • department
# Display all information.


class Person:
    def __init__(self, name):
        self.name = name
class Employee(Person):
    def __init__(self, name, employee_id):
            super().__init__(name)
            self.employee_id = employee_id
class Manager(Employee):
    def __init__(self, name, employee_id,department):
            super().__init__(name,employee_id)
            self.department = department
    def display(self):
            print("Name:", self.name)
            print("Employee ID:", self.employee_id)
            print("Department:",self.department)
#display
manager1 = Manager("Rohan",101, "Data Science")

manager1.display()
```

```python
# 17. Class Name:
# Animal → Mammal → Dog
# Problem
# Create multi-level inheritance.
# Explanation
# Animal
# • eat()
# ↓
# Mammal
# • walk()
# ↓
# Dog
# • bark()
# Demonstrate all inherited methods.

class Animal:
    def eat(self):
        print("Animal is eating")

class Mammal(Animal):
    def walk(self):
        print("Mammal is walking")
class Dog(Mammal):
    def bark(self):
        print("Dog is barking")
#display()
dog1 = Dog()
dog1.eat()
dog1.walk()
dog1.bark()
```

```python
# 19. Class Name:
# Shape → Polygon → Rectangle
# Problem
# Create multi-level inheritance.
# Explanation

# Sumit Sir Notes
# Rectangle should calculate its area using length and width.

class Shape:
    def display_shape(self):
        print("This is a shape")
class Polygon(Shape):
    def display_polygon(self):
        print("This is a polygon")
class Rectangle(Polygon):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        result = self.length * self.width

        print("Length:", self.length)
        print("Width:", self.width)
        print("Area of Rectangle:", result)


# Create Rectangle object
rectangle1 = Rectangle(10, 5)

# Demonstrate inherited methods
rectangle1.display_shape()
rectangle1.display_polygon()

# Calculate area
rectangle1.area()
```

```python
# 20. Class Name:
# BankAccount → SavingsAccount → RecurringDeposit
# Problem
# Create three levels of inheritance.
# Explanation
# Each class should add new functionality.
# Example:
# • deposit
# • interest
# • recurring deposit maturity

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        print("Current Balance:", self.balance)


class SavingsAccount(BankAccount):
    def calculate_interest(self, rate):
        interest = self.balance * rate / 100
        print("Interest:", interest)
        return interest


class RecurringDeposit(SavingsAccount):
    def calculate_maturity(self, monthly_deposit, months, rate):
        total_deposit = monthly_deposit * months

        # Simple interest approximation for demonstration
        interest = (
            monthly_deposit
            * months
            * (months + 1)
            / 2
            * rate
            / (12 * 100)
        )

        maturity_amount = total_deposit + interest

        print("Monthly Deposit:", monthly_deposit)
        print("Duration:", months, "months")
        print("Total Deposit:", total_deposit)
        print("Interest:", round(interest, 2))
        print("Maturity Amount:", round(maturity_amount, 2))


# Create RecurringDeposit object
rd = RecurringDeposit(5000)

# Level 1: BankAccount functionality
rd.deposit(2000)

# Level 2: SavingsAccount functionality
rd.calculate_interest(5)

# Level 3: RecurringDeposit functionality
rd.calculate_maturity(5000, 12, 6)
```

```python
# 22. Class Name:

# Sumit Sir Notes
# Product → Electronics → Smartphone
# Problem
# Create multi-level inheritance.
# Explanation
# Product
# • name
# • price
# Electronics
# • warranty
# Smartphone
# • RAM
# • Storage
# Display all specifications.

# 23. Class Name:
# LivingBeing → Animal → Bird
# Problem
# Create multi-level inheritance.
# Explanation
# Override methods at each level to demonstrate polymorphism.

class Product:
    def __init__(self, name,price):
        self.name = name
        self.price = price
class Electronics(Product):
    def __init__(self,name,price, warranty):
        super().__init__(name,price)
        self.warranty = warranty
class Smartphone(Electronics):
    def __init__(self, name, price, warranty, ram,storage):
        super().__init__(name,price, warranty)
        self.ram = ram
        self.storage = storage
    def display(self):
        print("Product Name:",self.name)
        print("Price:",self.price)
        print("Warranty:",self.warranty,"years")
        print("RAM:",self.ram)
        print("Storage:",self.storage)
#display
phone1 = Smartphone("Samsung Galaxy S25", 79999,2,"12 GB","256 GB")

phone1.display()
```
