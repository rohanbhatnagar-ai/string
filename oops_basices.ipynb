```python
# Class & Object

# Q1. Student Class
# Create a Student class with:

# name
# age
# course

# Create 2 objects and display their details.

# Concepts: class, __init__(), self, object creation, instance attributes.
class Student:
    def __init__(self,name,age,course):
        self.name = name
        self.age = age
        self.course = course
    # display details
    def display(self):
        print("name:",self.name)
        print("Age:",self.age)
        print("coursi:",self.course)
# display objects
s1 = Student("rohan",21,"CSE(AI)")
# display details
s1.display()
        
```

```python
# Q2. Employee with Method
# Create an Employee class with:

# emp_id
# name
# salary

# Create 3 objects and add a method display() to print employee details.

# Concepts: instance attributes, methods, multiple objects
class Employee:
    def __init__(self,emp_id,name,salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary
    # display 
    def display(self):
        print("employee_ID:",self.emp_id)
        print("employee_name:",self.name)
        print("employee_salary:",self.salary)

# creating
emp1 = Employee("43432","sumit sir",900000)
emp2 = Employee("34445","priya",800000)
# display details
emp1.display()
emp2.display()
        
```

```python
# Q3. Rectangle Area
# Create a Rectangle class with:

# length
# width

# Add methods:

# area()
# perimeter()

# Create an object and call both methods.

# Concepts: methods, self, calculations inside classes.

class Rectangle:
    def __init__(self,length,breadth):
        
        self.length = length
        self.breadth = breadth
    # def area()
    def area(self):
        print("area:",self.length*self.breadth)
    # def peramneters()
    def perimeters(self):
        print("perimeters:",self.length+self.breadth)
rec1 = Rectangle(10,5)
# display details
rec1.area()
rec1.perimeters()
```

```python
# Q4. __str__() Practice
# Create a Mobile class with:

# brand
# price
# color

# Implement __str__() so that:
class Mobile:
    def __init__(self,brand,price,color):
        self.brand = brand
        self.price = price
        self.color = color
    # def _str_():
    def _str_(self):
        print("brand:",self.brand)
        print("price:",self.price)
        print("color:",self.color)
#details
M1 = Mobile("Samsung",21000,"blue")
M2 = Mobile("Vivo",10000,"light white")
# display details
M1._str_()
```

```python
# Q10. Private Product Price
# Create a Product class with private:

# __price

# Add:

# display_price()

# Then try changing/accessing __price directly from outside.

# Concepts: data hiding, private attributes.
class Products:
    def __init__(self,price):
        self._price = price
    # def display
    def display_price(self):
        print("display_price",self._price)
# details
P1= Products(600000)
P1.display_price()
```

```python
# Getter & Setter

# Q11. Student Marks — Getter & Setter
# Create a Student class with private:

# __marks

# Create:

# get_marks()
# set_marks()

# The setter should accept marks only between 0 and 100.

# Example:

# Marks: 85
# Invalid marks

class Student:
    def __init__(self,marks):
        self._marks=marks
    # getter
    def get_marks(self):
        return self._marks
    # setter
    def set_marks(self,marks):
        if 0<=marks<=100:
            print("marks",self._marks)
# details
S1 = Student(1000)
# get marks
S1.get_marks
# setters
S1.set_marks(100)
```

```python
# Q12. Bank Account — Balance Validation
# Create a BankAccount class with private:

# __balance

# Implement:

# get_balance()
# set_balance()

# Setter should:

# Reject negative balance.
# Accept zero or positive values.

# Concepts: getter/setter + validation + encapsulation.

class BankAccount:
    def __init__(self,balance):
        self._balance = balance
    # get the balance
    def get_balance(self):
        return get_balance
    # setter
    def set_balance(self,balance):
        self._balance = balance
        if balance>0:
            print("positive number")
        else:
            print("negative")
# details
Bank1 = BankAccount(4000)
# # get the methods
# Bank1.get_balance()
# setter
Bank1.set_balance(5000)
            
```
